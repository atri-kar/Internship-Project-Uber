```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
```

## 1. Load both datasets

```python
traffic = pd.read_csv("/content/Dataset_Uber Traffic.csv")
weather = pd.read_csv("/content/Bangalore Weather Data 2015-11-01 to 2017-06-30.csv")
events = pd.read_csv("/content/Bangalore_Special_Events_2015_2017.csv")
```

## 2. Fixing date formats

```python
# Parse datetime
traffic["DateTime"] = pd.to_datetime(traffic["DateTime"], format="%d/%m/%y %H:%M", dayfirst=True)

# Create Date + Hour
traffic["Date"] = traffic["DateTime"].dt.date
traffic["Hour"] = traffic["DateTime"].dt.hour

# Sort
traffic = traffic.sort_values(by=["Junction", "DateTime"])
```

## 3. Clean and Prepare Important Weather Data

```python
weather["Date"] = pd.to_datetime(weather["datetime"], format="%d/%m/%Y").dt.date

# Dropping unwanted columns
weather_clean = weather[['Date', 'temp', 'humidity', 'precip', 'windspeed', 'cloudcover', 'visibility', 'conditions']]
```

## 4. Merge Hourly Traffic Data wth Weather Data

```python
merged = traffic.merge(weather_clean, on="Date", how="left")
```

```python
# Checking the head and shape of the merged dataset
print(merged.head())
print(merged.shape)
```

## 5. Clean Events Data and Merge

```python
events['Date'] = pd.to_datetime(events['date']).dt.date
events_clean = events.drop(columns=['date'])
```

```python
# Merging events_clean with merged dataset
merged = merged.merge(events_clean, on = 'Date', how = 'left')
```

```python
merged.head()
```

```python
# Fill non-event days because null values are due to 'None' category
merged["event_name"] = merged["event_name"].fillna("No Event")
merged["category"] = merged["category"].fillna("No Category")
merged["impact_level"] = merged["impact_level"].fillna("No impact")
```
